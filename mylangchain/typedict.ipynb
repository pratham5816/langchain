```python
### data classes
from langchain.agents import create_agent
from dataclasses import dataclass
from langchain_ollama import ChatOllama
from langchain.messages import SystemMessage,HumanMessage

@dataclass
class Car:
    """The detailed format of car"""
    name:str
    year:int
    cc:int


model = ChatOllama(
    model="qwen3:4b",
    temperature=0
)

agent = create_agent(
    model=model,
    response_format=Car
)

messages = [
    HumanMessage(content="Give details of bmw m5")  
]

messages = [
    SystemMessage(content=(
        "You are a BMW engineer and technical assistant. "
        "Answer concisely and accurately. "
        "Use available tools when needed. "
        "Never invent technical data."
    )),  
    HumanMessage(content="Give details of bmw m5")  
]

resp = agent.invoke({"messages":messages})


print(resp)
```

```python
### compare to pydantic typedict dont have validations of output. ideal when validation is not required

from typing_extensions import TypedDict , Annotated
from langchain_ollama import ChatOllama
from langchain.messages import SystemMessage,HumanMessage


model = ChatOllama(
    model="llama3.2:1b",
    temperature=0
)

class Bike(TypedDict):
    """Detailed output for a bike"""
    name: Annotated[str, ... ,"The name of the bike"]
    brand: Annotated[str, ... , "The brand of the bike"]
    cc: Annotated[int, ... , "The engine cc of the bike"]

model_with_fields=model.with_structured_output(Bike)
    

model_with_fields.invoke("Give the detail of bike called as z900 kawasaki")


```

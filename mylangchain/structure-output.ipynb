```python
### nested structure
from pydantic import Field
class Bike(BaseModel):
    name:str=Field(description="The name of the bike")
    company:str=Field(description="The name of the brand")
    cc:int=Field(description="The cc of bike's engine")

class Bikes(BaseModel):
    company:str=Field(description="The name of the brand")
    bikes:list[Bike]=Field(description="The list of bikes")


model_with_fields_list = model.with_structured_output(Bikes)

model_with_fields_list.invoke("""
    Give me exactly five Kawasaki bikes.

    Set the outer company field to "Kawasaki".
    For each bike, provide:
    - name
    - company
    - engine displacement in cc

    Return all five bikes in the bikes list.
    """)
```

```python
from langchain.messages import SystemMessage , HumanMessage , AIMessage 
from langchain_ollama import ChatOllama

model = ChatOllama(
    model="llama3.2:1b",
    temperature=0
)


messages = [
    SystemMessage("You are an experienced Kawasaki motorcycle engineer."),    
    HumanMessage("Why does an inline four cylinder engine heats to much"),
    AIMessage("An overheating issue can have several causes, including low coolant...")
]

res = model.invoke(messages)
print(res.content)
```

```python
from langchain.messages import SystemMessage , HumanMessage , AIMessage 
from langchain_ollama import ChatOllama

## thier are four types of messages -> system message , human message , tool message and ai message.
model = ChatOllama(
    model="llama3.2:1b",
    temperature=0
)

sm = SystemMessage("""
You are an experienced motorcycle engineer and passionate biker specializing in Kawasaki motorcycles.

Your responsibilities:
- Provide accurate and practical information about Kawasaki motorcycles.
- Explain motorcycle mechanics, engines, transmissions, suspension, braking, electronics, maintenance, and troubleshooting.
- Help users understand Kawasaki models, specifications, features, and riding characteristics.
- Give engineering explanations in a clear and easy-to-understand way.

Communication style:
- Be concise but technically informative.
- Speak like an experienced motorcycle engineer explaining something to another biker.
- Use examples when they make the explanation easier to understand.
""")

messages = [
    sm,
    HumanMessage("What is an inline four cylinder engine")
]

r = model.invoke(messages)
r.content
```
